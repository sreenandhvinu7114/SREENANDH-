```python
df["date"] = pd.to_datetime(df["date"])
```

```python

plt.figure(figsize=(12,5))
sns.lineplot(
    data=df,
    x=df.index,
    y="close"
)
plt.title("shopify closing price over time")
plt.xlabel("date")
plt.ylabel("closing price")

plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.lineplot(
    data=df,
    x=df.index,
    y="open"
)
plt.title("Shopify Opening Price Over Time")
plt.xlabel("Date")
plt.ylabel("Opening Price")
plt.show()
```

```python
#visualize OHLC price
plt.figure(figsize=(14,6))

sns.lineplot(data=df,x=df.index,y="open",label="open")
sns.lineplot(data=df,x=df.index,y="high",label="high")
sns.lineplot(data=df,x=df.index,y="low",label="low")
sns.lineplot(data=df,x=df.index,y="close",label="close")

plt.title("shopify OHLC prices over time")
plt.xlabel("date")
plt.ylabel("price")

plt.show()
```

```python
#visulalizing trading volume over time
plt.figure(figsize=(14,5))
sns.lineplot(
    data=df,
    x="date",
    y="volume"

    )
plt.title("shopify trading volume over time")
plt.xlabel("date")
plt.ylabel("trading volume")

```

```python
df["daily price change"] = df["close"] - df["open"]
df["daily price change"]
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price change"
)
plt.title("Trading volume vs daily price change")
plt.xlabel("volume")
plt.ylabel("Daily Price Change")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="volume",
    y="daily price change"
)
plt.title("Volume vs daily price change")
plt.xlabel("volume")
plt.ylabel("Daily Price Change")

plt.show()
```

```python
#calculate the 20 days\
df["SMA_20"]= df['close'].rolling(window=20).mean()
df["SMA_20"]
```

```python
#Plot Closing Pricing + 20 Day SMA
plt.figure(figsize=(14,6))
sns.lineplot(
    data=df,
    x=df.index,
    y="close",
    label="Closing Price")

sns.lineplot(
    data=df,
    x=df.index,
    y="SMA_20",
    label="20-Day SMA")

plt.title("Shopify Closing Price and 20-Day SMA")
plt.xlabel("Date")
plt.ylabel("Price")

plt.show()

```

```python
#Calculate the 20-Day SMA


df["SMA_50"] = df["close"].rolling(window=50).mean()
df["SMA_50"]
```

```python
#plot closing price + 50-Day SMA

plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x="date",
    y="close",
    label="Closing price"
)

sns.lineplot(
    data=df,
    x="date",
    y="SMA_50",
    label="50-Day SMA"
)

plt.title("Shopify Closing Price and 50-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("SMA_50")

plt.show()
```

```python
#plot Close + Both Moving Average
#Calculate the 20-Day SMA

df["SMA_20"] = df["close"].rolling(window=20).mean()
df["SMA_20"]
plt.figure(figsize=(14,6))

sns.lineplot(
    data=df,
    x="date",
    y="close",
    label="Closing Price"
)

sns.lineplot(
    data=df,
    x="date",
    y="SMA_20",
    label="20-Day SMA"
)

plt.title("Shopify Closing Price and 20-Day Moving Average")
plt.xlabel("Date")
plt.ylabel("SMA_20")

plt.show()
```

```python
#daily return %
df["daily return %"] = ((df["close"]) - (df["open"])) / (df["open"]) * 100
```

```python
df["daily return %"].head()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="daily return %",
    bins=20,
    kde=True
)

plt.title("Distribution of Daily Returns")
plt.xlabel("Daily Return %")
plt.ylabel("Frequency")

plt.show()
```
